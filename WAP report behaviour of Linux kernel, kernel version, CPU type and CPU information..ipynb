```python
import platform
import subprocess
 
def get_kernel_version():
    """Get the Linux kernel version."""
    return platform.uname().release
 
def get_cpu_type():
    """Get the CPU type (architecture)."""
    return platform.uname().machine
 
def get_cpu_info():
    """Get detailed CPU information."""
    try:
        cpu_info = subprocess.check_output("lscpu", shell=True).decode('utf-8')
        return cpu_info
    except Exception as e:
        return f"Error retrieving CPU information: {e}"
 
if __name__ == "__main__":
    print("Linux Kernel Version:", get_kernel_version())
    print("CPU Type:", get_cpu_type())
    print("Detailed CPU Information:\n", get_cpu_info())
```
