```python
class Process:
    def __init__(self, pid, burst_time, priority):
        self.pid = pid
        self.burst_time = burst_time
        self.priority = priority
        self.waiting_time = 0
        self.turnaround_time = 0

def calculate_waiting_time(processes):
    processes[0].waiting_time = 0
    for i in range(1, len(processes)):
        processes[i].waiting_time = processes[i - 1].waiting_time + processes[i - 1].burst_time

def calculate_turnaround_time(processes):
    for process in processes:
        process.turnaround_time = process.burst_time + process.waiting_time

def priority_scheduling(processes):
    processes.sort(key=lambda x: (x.priority, x.pid))
    calculate_waiting_time(processes)
    calculate_turnaround_time(processes)
    print("Processes  Burst Time  Priority  Waiting Time  Turnaround Time")
    total_waiting_time = 0
    total_turnaround_time = 0
    for p in processes:
        total_waiting_time += p.waiting_time
        total_turnaround_time += p.turnaround_time
        print(f"   {p.pid}\t\t{p.burst_time}\t\t{p.priority}\t\t{p.waiting_time}\t\t{p.turnaround_time}")
    print(f"\nAverage waiting time = {total_waiting_time / len(processes):.2f}")
    print(f"Average turnaround time = {total_turnaround_time / len(processes):.2f}")

if __name__ == "__main__":
    processes = [
        Process(1, 10, 3),
        Process(2, 1, 1),
        Process(3, 2, 4),
        Process(4, 1, 5),
        Process(5, 5, 2),
    ]
    priority_scheduling(processes)
```
