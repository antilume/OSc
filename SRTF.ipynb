```python
import sys

def find_waiting_time(processes, n, burst_time, waiting_time, arrival_time):
    remaining_time = [0] * n
    for i in range(n):
        remaining_time[i] = burst_time[i]
    complete = 0
    current_time = 0
    min_time = sys.maxsize
    shortest = 0
    check = False
    while complete != n:
        for j in range(n):
            if (arrival_time[j] <= current_time and
                remaining_time[j] > 0 and
                remaining_time[j] < min_time):
                min_time = remaining_time[j]
                shortest = j
                check = True
        if check is False:
            current_time += 1
            continue
        remaining_time[shortest] -= 1
        min_time = remaining_time[shortest]
        if min_time == 0:
            min_time = sys.maxsize
        if remaining_time[shortest] == 0:
            complete += 1
            check = False
            finish_time = current_time + 1
            waiting_time[shortest] = finish_time - burst_time[shortest] - arrival_time[shortest]
            if waiting_time[shortest] < 0:
                waiting_time[shortest] = 0
        current_time += 1

def find_turnaround_time(processes, n, burst_time, waiting_time, turnaround_time):
    for i in range(n):
        turnaround_time[i] = burst_time[i] + waiting_time[i]

def find_avg_time(processes, n, burst_time, arrival_time):
    waiting_time = [0] * n
    turnaround_time = [0] * n
    find_waiting_time(processes, n, burst_time, waiting_time, arrival_time)
    find_turnaround_time(processes, n, burst_time, waiting_time, turnaround_time)
    print("Processes    Burst Time    Arrival Time    Waiting Time    Turnaround Time")
    total_waiting_time = 0
    total_turnaround_time = 0
    for i in range(n):
        total_waiting_time += waiting_time[i]
        total_turnaround_time += turnaround_time[i]
        print(f"   {i+1}\t\t{burst_time[i]}\t\t{arrival_time[i]}\t\t{waiting_time[i]}\t\t{turnaround_time[i]}")
    print(f"\nAverage waiting time = {total_waiting_time / n:.2f}")
    print(f"Average turnaround time = {total_turnaround_time / n:.2f}")

if __name__ == "__main__":
    processes = ['P1', 'P2', 'P3', 'P4']
    n = len(processes)
    burst_time = [6, 8, 7, 3]
    arrival_time = [0, 1, 2, 3]
    find_avg_time(processes, n, burst_time, arrival_time)
```
