```python
import os
import shutil
 
# Method 1: Using os.system("cp ...")
def copy_file_system(source, destination):
    """Copy a file using the os.system() system call."""
    exit_code = os.system(f"cp {source} {destination}")
    if exit_code == 0:
        print(f"File copied from {source} to {destination}")
    else:
        print(f"Error copying file: exit code {exit_code}")
 
# Method 2: Using shutil.copy()
def copy_file_shutil(source, destination):
    """Copy a file using the shutil.copy() function."""
    try:
        shutil.copy(source, destination)
        print(f"File copied from {source} to {destination}")
    except FileNotFoundError:
        print(f"Error: Source file '{source}' not found.")
    except PermissionError:
        print(f"Error: Permission denied.")
 
if __name__ == "__main__":
    # Method 1: os.system()
    src = input("Enter source file name: ")
    dst = input("Enter destination file name: ")
    copy_file_system(src, dst)
 
    # Method 2: shutil.copy()
    src = input("Enter source file name: ")
    dst = input("Enter destination file name: ")
    copy_file_shutil(src, dst)
```
