```python
import multiprocessing
import os
import time


# ==========================================
# 1: Parent & Child Running Same Code
# ==========================================

def same_task():
    print(f"[PROCESS] PID: {os.getpid()} — running the same task")


def example_same_code():
    print(f"[PARENT] PID: {os.getpid()} — starting")

    child = multiprocessing.Process(target=same_task)

    child.start()

    # Parent also executes the same function
    same_task()

    child.join()

    print(f"[PARENT] PID: {os.getpid()} — done")


# ==========================================
# 2: Parent & Child Running Different Code
# ==========================================

def child_specific_task():
    print(f"[CHILD] PID: {os.getpid()} — child-specific work")

    result = sum(range(1, 11))

    print(f"[CHILD] Computed sum 1..10 = {result}")
    print(f"[CHILD] Exiting")


def parent_specific_task():
    print(f"[PARENT] PID: {os.getpid()} — parent-specific work")
    print(f"[PARENT] Spawning child process...")


def example_different_code():
   

    parent_specific_task()

    child = multiprocessing.Process(target=child_specific_task)

    child.start()
    child.join()

    print(f"[PARENT] PID: {os.getpid()} — child finished")
    print(f"[PARENT] Parent exiting")


# ==========================================
# 3: Waiting for Child Using join()
# ==========================================

def long_running_child_task():
    print(f"[CHILD] PID: {os.getpid()} — starting long task")

    time.sleep(2)

    print(f"[CHILD] Task complete")
    print(f"[CHILD] Exiting")


def example_join_and_exitcode():
    print(f"[PARENT] PID: {os.getpid()} — creating child process")

    child = multiprocessing.Process(target=long_running_child_task)

    child.start()

    print(f"[PARENT] Child spawned with PID {child.pid}")
    print(f"[PARENT] Waiting for child to finish...")

    child.join()

    print(f"[PARENT] Child exit code: {child.exitcode}")
    print(f"[PARENT] Parent exiting")




if __name__ == "__main__":
    example_same_code()
    example_different_code()
    example_join_and_exitcode()
```

```python
import multiprocessing
import os
import time


def same_task():
    print(f"[PROCESS] PID: {os.getpid()} — running the same task")


def same_code():
    print("\n========== 1 ==========")
    print(f"[PARENT] PID: {os.getpid()} — starting")

    child = multiprocessing.Process(target=same_task)

    child.start()

    same_task()

    child.join()

    print(f"[PARENT] PID: {os.getpid()} — done")


def child_specific_task():
    print(f"[CHILD] PID: {os.getpid()} — child-specific work")

    result = sum(range(1, 11))

    print(f"[CHILD] Computed sum 1..10 = {result}")
    print(f"[CHILD] Exiting")


def parent_specific_task():
    print(f"[PARENT] PID: {os.getpid()} — parent-specific work")
    print(f"[PARENT] Spawning child process...")


def different_code():
    print("\n========== 2 ==========")

    parent_specific_task()

    child = multiprocessing.Process(target=child_specific_task)

    child.start()
    child.join()

    print(f"[PARENT] PID: {os.getpid()} — child finished")
    print(f"[PARENT] Parent exiting")


def long_running_child_task():
    print(f"[CHILD] PID: {os.getpid()} — starting long task")

    time.sleep(2)

    print(f"[CHILD] Task complete")
    print(f"[CHILD] Exiting")


def join_and_exitcode():
    print("\n========== 3 ==========")
    print(f"[PARENT] PID: {os.getpid()} — creating child process")

    child = multiprocessing.Process(target=long_running_child_task)

    child.start()

    print(f"[PARENT] Child spawned with PID {child.pid}")
    print(f"[PARENT] Waiting for child to finish...")

    child.join()

    print(f"[PARENT] Child exit code: {child.exitcode}")
    print(f"[PARENT] Parent exiting")


if __name__ == "__main__":
    same_code()
    different_code()
    join_and_exitcode()
```
