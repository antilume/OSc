```python
def find_waiting_time(processes, n, burst_time, waiting_time):
    waiting_time[0] = 0
    for i in range(1, n):
        waiting_time[i] = burst_time[i - 1] + waiting_time[i - 1]
 
def find_turnaround_time(processes, n, burst_time, waiting_time, turnaround_time):
    for i in range(n):
        turnaround_time[i] = burst_time[i] + waiting_time[i]
 
def find_avg_time(processes, n, burst_time):
    waiting_time = [0] * n
    turnaround_time = [0] * n
    find_waiting_time(processes, n, burst_time, waiting_time)
    find_turnaround_time(processes, n, burst_time, waiting_time, turnaround_time)
    print("Processes    Burst Time    Waiting Time    Turnaround Time")
    total_waiting_time = 0
    total_turnaround_time = 0
    for i in range(n):
        total_waiting_time += waiting_time[i]
        total_turnaround_time += turnaround_time[i]
        print(f"   {i+1}\t\t{burst_time[i]}\t\t{waiting_time[i]}\t\t{turnaround_time[i]}")
    print(f"\nAverage waiting time = {total_waiting_time / n:.2f}")
    print(f"Average turnaround time = {total_turnaround_time / n:.2f}")
 
if __name__ == "__main__":
    processes = ['P1', 'P2', 'P3', 'P4']
    n = len(processes)
    burst_time = [5, 9, 6, 2]
    find_avg_time(processes, n, burst_time)
```
